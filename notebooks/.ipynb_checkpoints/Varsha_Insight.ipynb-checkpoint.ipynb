# model building

```python
# ================================
# CREATE MONTHLY RAINFALL SERIES
# ================================

# Month columns
months = ['JAN','FEB','MAR','APR','MAY','JUN',
          'JUL','AUG','SEP','OCT','NOV','DEC']

# Convert wide dataset to long format
monthly_data = df.melt(
    id_vars=['YEAR'],
    value_vars=months,
    var_name='MONTH',
    value_name='RAINFALL'
)

# Month mapping
month_map = {
    'JAN':1,'FEB':2,'MAR':3,'APR':4,
    'MAY':5,'JUN':6,'JUL':7,'AUG':8,
    'SEP':9,'OCT':10,'NOV':11,'DEC':12
}

# Convert month names to numbers
monthly_data['MONTH_NUM'] = monthly_data['MONTH'].map(month_map)

# Create DATE column
monthly_data['DATE'] = pd.to_datetime(
    monthly_data['YEAR'].astype(str)
    + '-'
    + monthly_data['MONTH_NUM'].astype(str)
)

# Set date index
monthly_data.set_index('DATE', inplace=True)

# Average rainfall across India
monthly_rainfall = monthly_data.groupby('DATE')['RAINFALL'].mean()

# Display
print(monthly_rainfall.head())
```

```python
# ================================
# TRAIN TEST SPLIT
# ================================

# Training data
train = monthly_rainfall[:'2009']

# Testing data
test = monthly_rainfall['2010':]

print(train.shape)
print(test.shape)
```

# Time Series Forecasting using SARIMA

```python
# ================================
# SARIMA MODEL BUILDING
# ================================

from statsmodels.tsa.statespace.sarimax import SARIMAX

model = SARIMAX(
    train,
    order=(1,1,1),
    seasonal_order=(1,1,1,12)
)

sarima_model = model.fit()

print(sarima_model.summary())
```

```python
# ================================
# VALIDATION PREDICTION
# ================================

predictions = sarima_model.predict(
    start=len(train),
    end=len(train)+len(test)-1
)
```

```python
# ================================
# MODEL EVALUATION
# ================================

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
import numpy as np

mae = mean_absolute_error(test, predictions)

mse = mean_squared_error(test, predictions)

rmse = np.sqrt(mse)

accuracy = 100 - (mae / test.mean()) * 100

print("MAE :", mae)

print("MSE :", mse)

print("RMSE :", rmse)

print("Forecast Accuracy :", accuracy)
```

```python
# ================================
# ACTUAL VS PREDICTED GRAPH
# ================================

import matplotlib.pyplot as plt

plt.figure(figsize=(18,6))

# Actual rainfall
plt.plot(
    test.index,
    test.values,
    label='Actual Rainfall',
    color='blue'
)

# Predicted rainfall
plt.plot(
    test.index,
    predictions,
    label='Predicted Rainfall',
    color='orange'
)

plt.title("Monthly Rainfall Forecast Validation")

plt.xlabel("Year")

plt.ylabel("Rainfall")

plt.legend()

plt.grid(True)

plt.show()
```

# Predict Future Rainfall

```python
# ================================
# FUTURE FORECASTING
# ================================

# Forecast next 120 months
future_forecast = sarima_model.forecast(steps=120)

# Future dates
future_dates = pd.date_range(
    start=monthly_rainfall.index[-1] + pd.DateOffset(months=1),
    periods=120,
    freq='MS'
)

# Forecast dataframe
forecast_df = pd.DataFrame({
    'Date': future_dates,
    'Forecasted_Rainfall': future_forecast.values
})

print(forecast_df.head())
```

```python
# ================================
# FUTURE FORECAST GRAPH
# ================================

plt.figure(figsize=(18,6))

# Historical rainfall
plt.plot(
    monthly_rainfall.index,
    monthly_rainfall.values,
    label='Historical Rainfall',
    color='blue'
)

# Forecast rainfall
plt.plot(
    forecast_df['Date'],
    forecast_df['Forecasted_Rainfall'],
    label='Forecasted Rainfall',
    color='red'
)

plt.title("Future Monthly Rainfall Forecast")

plt.xlabel("Year")

plt.ylabel("Rainfall")

plt.legend()

plt.grid(True)

plt.show()
```

```python
# ================================
# MONTH-WISE ANOMALY DETECTION
# ================================

# Extract month number
forecast_df['Month'] = forecast_df['Date'].dt.month

# Historical monthly averages
monthly_avg = monthly_rainfall.groupby(
    monthly_rainfall.index.month
).mean()

# Function to classify rainfall
def classify_rainfall(row):

    month_avg = monthly_avg[row['Month']]

    if row['Forecasted_Rainfall'] < month_avg * 0.75:
        return 'Drought'

    elif row['Forecasted_Rainfall'] > month_avg * 1.25:
        return 'Heavy Rainfall'

    else:
        return 'Normal'

# Apply classification
forecast_df['Forecast_Status'] = forecast_df.apply(
    classify_rainfall,
    axis=1
)

# Display
print(forecast_df.head())
```

```python
# ================================
# EXPORT FORECAST CSV
# ================================

forecast_df.to_csv(
    "monthly_rainfall_forecast.csv",
    index=False
)

print("Forecast CSV Exported Successfully")
```
