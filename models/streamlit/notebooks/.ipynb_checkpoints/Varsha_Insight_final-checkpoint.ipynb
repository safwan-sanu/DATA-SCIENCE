# Drought & Heavy Rainfall Detection

```python
# Historical Average Rainfall
average_rainfall = df['ANNUAL'].mean()

# Rainfall Classification Function
def rainfall_status(rainfall):

    if rainfall < average_rainfall * 0.60:
        return "Severe Drought Risk"

    elif rainfall < average_rainfall * 0.80:
        return "Drought Warning"

    elif rainfall < average_rainfall * 0.95:
        return "Below Normal Rainfall"

    elif rainfall <= average_rainfall * 1.10:
        return "Normal Rainfall"

    elif rainfall <= average_rainfall * 1.30:
        return "Heavy Rainfall Alert"

    else:
        return "Flood Risk / Extreme Rainfall"

# Create Status Column
df['RAINFALL_STATUS'] = df['ANNUAL'].apply(
    rainfall_status
)

df[['YEAR', 'ANNUAL', 'RAINFALL_STATUS']].head()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    x='RAINFALL_STATUS',
    data=df,
    order=df['RAINFALL_STATUS'].value_counts().index
)

plt.title("Rainfall Status Classification")

plt.xlabel("Rainfall Status")

plt.ylabel("Count")

plt.xticks(rotation=20)

plt.show()
```

# model building

```python
# Yearly Rainfall Time Series
yearly_rainfall = df.groupby('YEAR')['ANNUAL'].mean()

# Convert index to datetime
yearly_rainfall.index = pd.to_datetime(
    yearly_rainfall.index,
    format='%Y'
)

yearly_rainfall
```

```python
plt.figure(figsize=(15,6))

plt.plot(
    yearly_rainfall.index,
    yearly_rainfall.values,
    marker='o'
)

plt.title("Yearly Rainfall Trend")

plt.xlabel("Year")

plt.ylabel("Average Rainfall")

plt.grid(True)

plt.show()
```

# model building

```python
train = yearly_rainfall.iloc[:-10]

test = yearly_rainfall.iloc[-10:]

print("Train Size:", len(train))
print("Test Size:", len(test))
```

```python
from statsmodels.tsa.arima.model import ARIMA
import warnings

warnings.filterwarnings("ignore")

best_aic = float("inf")
best_order = None

for p in range(0, 4):
    for d in range(0, 3):
        for q in range(0, 4):

            try:
                model = ARIMA(
                    train,
                    order=(p,d,q)
                )

                results = model.fit()

                if results.aic < best_aic:
                    best_aic = results.aic
                    best_order = (p,d,q)

            except:
                continue

print("Best ARIMA Order:", best_order)
print("Best AIC:", best_aic)
```

```python
model = ARIMA(
    train,
    order=best_order
)

results = model.fit()

print(results.summary())
```

```python
predictions = results.forecast(
    steps=len(test)
)

predictions.index = test.index

predictions
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

import numpy as np

mae = mean_absolute_error(
    test,
    predictions
)

rmse = np.sqrt(
    mean_squared_error(
        test,
        predictions
    )
)

mape = np.mean(
    np.abs((test - predictions) / test)
) * 100

accuracy = 100 - mape

print("MAE :", round(mae,2))

print("RMSE :", round(rmse,2))

print("MAPE :", round(mape,2), "%")

print("Accuracy :", round(accuracy,2), "%")
```

```python
plt.figure(figsize=(15,6))

plt.plot(
    train.index,
    train,
    label='Train Data'
)

plt.plot(
    test.index,
    test,
    label='Actual Rainfall',
    marker='o'
)

plt.plot(
    predictions.index,
    predictions,
    label='Predicted Rainfall',
    marker='o'
)

plt.title('Actual vs Predicted Rainfall')

plt.xlabel('Year')

plt.ylabel('Rainfall')

plt.legend()

plt.grid(True)

plt.show()
```

```python
#CHECK SARIMA
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
import warnings

warnings.filterwarnings("ignore")

sarima_model = SARIMAX(
    train,
    order=(1,1,1),
    seasonal_order=(1,1,1,12)
)

sarima_results = sarima_model.fit()

print(sarima_results.summary())
```

```python
sarima_predictions = sarima_results.forecast(
    steps=len(test)
)

sarima_predictions.index = test.index

sarima_predictions
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
import numpy as np

mae = mean_absolute_error(
    test,
    sarima_predictions
)

rmse = np.sqrt(
    mean_squared_error(
        test,
        sarima_predictions
    )
)

mape = np.mean(
    np.abs((test - sarima_predictions) / test)
) * 100

accuracy = 100 - mape

print("MAE :", round(mae,2))
print("RMSE :", round(rmse,2))
print("MAPE :", round(mape,2), "%")
print("Accuracy :", round(accuracy,2), "%")
```

```python
plt.figure(figsize=(15,6))

plt.plot(
    train.index,
    train,
    label="Train Data"
)

plt.plot(
    test.index,
    test,
    label="Actual Rainfall",
    marker="o"
)

plt.plot(
    sarima_predictions.index,
    sarima_predictions,
    label="SARIMA Prediction",
    marker="o"
)

plt.title("Actual vs Predicted Rainfall (SARIMA)")
plt.xlabel("Year")
plt.ylabel("Rainfall")
plt.legend()
plt.grid(True)

plt.show()
```

To identify the most suitable forecasting model, both ARIMA and SARIMA models were developed and evaluated using MAE, RMSE, MAPE, and Accuracy metrics. The ARIMA model achieved an accuracy of 90.48%, whereas the SARIMA model achieved an accuracy of 92.83%. Additionally, SARIMA produced lower MAE (92.83), RMSE (109.44), and MAPE (7.17%) values compared to ARIMA. Since SARIMA incorporates both trend and seasonal components of rainfall data, it was able to capture rainfall patterns more effectively. Therefore, SARIMA was selected as the final forecasting model for future rainfall prediction in the proposed system.


```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

final_model = SARIMAX(
    yearly_rainfall,
    order=(1,1,1),
    seasonal_order=(1,1,1,12)
)

final_results = final_model.fit()
```

```python
future_forecast = final_results.forecast(
    steps=10
)

future_years = pd.date_range(
    start=yearly_rainfall.index[-1],
    periods=11,
    freq='Y'
)[1:]

future_forecast.index = future_years

future_forecast
```

```python
plt.figure(figsize=(15,6))

plt.plot(
    yearly_rainfall.index,
    yearly_rainfall,
    label='Historical Rainfall'
)

plt.plot(
    future_forecast.index,
    future_forecast,
    label='SARIMA Forecast',
    marker='o'
)

plt.title("Future Rainfall Forecast (SARIMA)")

plt.xlabel("Year")
plt.ylabel("Rainfall")

plt.legend()
plt.grid(True)

plt.show()
```

```python
forecast_df = pd.DataFrame({
    'Year': future_forecast.index.year,
    'Forecasted_Rainfall': future_forecast.values
})

forecast_df
```

```python
# Historical Mean Rainfall
mean_rainfall = yearly_rainfall.mean()

print("Historical Mean Rainfall:", round(mean_rainfall,2))
```

```python
def rainfall_alert(rainfall, mean_rainfall):

    if rainfall < mean_rainfall * 0.60:
        return "Severe Drought Risk"

    elif rainfall < mean_rainfall * 0.80:
        return "Drought Warning"

    elif rainfall < mean_rainfall * 0.95:
        return "Below Normal Rainfall"

    elif rainfall <= mean_rainfall * 1.10:
        return "Normal Rainfall"

    elif rainfall <= mean_rainfall * 1.30:
        return "Heavy Rainfall Alert"

    else:
        return "Flood Risk / Extreme Rainfall"
```

```python
forecast_alerts = pd.DataFrame({
    'Year': future_forecast.index.year,
    'Forecasted_Rainfall': future_forecast.values
})

forecast_alerts['Alert'] = forecast_alerts[
    'Forecasted_Rainfall'
].apply(
    lambda x: rainfall_alert(x, mean_rainfall)
)

forecast_alerts
```

```python
plt.figure(figsize=(16,6))

plt.plot(+
    forecast_alerts['Year'],
    forecast_alerts['Forecasted_Rainfall'],
    marker='o'
)

for i in range(len(forecast_alerts)):

    plt.text(
        forecast_alerts['Year'][i],
        forecast_alerts['Forecasted_Rainfall'][i] + 10,
        forecast_alerts['Alert'][i],
        fontsize=8,
        rotation=45
    )

plt.axhline(
    mean_rainfall,
    color='red',
    linestyle='--',
    label='Historical Mean'
)

plt.title("Future Rainfall Risk Alerts")

plt.xlabel("Year")

plt.ylabel("Forecasted Rainfall")

plt.legend()

plt.grid(True)

plt.show()
```

```python
forecast_alerts.to_csv(
    "forecasted_rainfall.csv",
    index=False
)
```

The “Varsha Insight – Rainfall Analytics, Forecasting, and Climate Intelligence System” was successfully developed to analyze historical rainfall patterns and predict future rainfall trends across India using data analytics, machine learning, and big data technologies. The project utilized a comprehensive rainfall dataset containing monthly, seasonal, and annual rainfall records from various meteorological subdivisions of India. Exploratory Data Analysis (EDA), statistical analysis, and data visualization techniques were performed to identify rainfall trends, seasonal variations, and regional climate patterns.

To achieve accurate forecasting, both ARIMA and SARIMA time-series forecasting models were evaluated using performance metrics such as MAE, RMSE, and MAPE. The SARIMA model demonstrated superior forecasting accuracy and was therefore selected as the final prediction model. The model was used to forecast future rainfall patterns and generate meaningful climate insights for decision-making.

The project was further enhanced with a Rainfall Alert and Farmer Advisory System that categorizes forecasted rainfall into classes such as Severe Drought Risk, Drought Warning, Below Normal Rainfall, Normal Rainfall, Heavy Rainfall Alert, and Flood Risk. Based on these predictions, advisory recommendations are generated to support agricultural planning, irrigation management, water resource utilization, and disaster preparedness.

Interactive dashboards were developed using Streamlit, Tableau, and Power BI to visualize rainfall trends, forecasting results, and climate insights in a user-friendly manner. In addition, Hadoop MapReduce and Apache Spark were implemented to demonstrate distributed processing and large-scale rainfall data analytics. Overall, the project successfully integrates predictive analytics, climate intelligence, visualization tools, and big data technologies to provide an effective decision-support system for rainfall monitoring and forecasting.
